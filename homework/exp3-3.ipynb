```python
# （1）导入库
from sklearn.datasets import load_breast_cancer
from sklearn.neighbors import KNeighborsClassifier  # 导入KNN分类库
from sklearn.model_selection import train_test_split  # 导入数据集切分库
from sklearn.metrics import classification_report
```

```python
# （2）调用load_breast_cancer()读入信息
digits = load_breast_cancer()
```

```python
# （3）分割数据,训练集和测试集为3：1
X_train, X_test, Y_train, Y_test = train_test_split(
    digits.data, digits.target, test_size=0.25, random_state=33
)  # type: ignore
```

```python
# （4）利用KNN模型进行分类
knc = KNeighborsClassifier()
knc.fit(X_train, Y_train)
y_predict = knc.predict(X_test)  # 预测测试集
```

```python
# （5）使用模型自带的评估函数进行准确性测评。
print("The accuracy of K-Nearest Neighbor Classifier is", knc.score(X_test, Y_test))
```

```python
# （6）使用sklearn.metrics里面的classification_report模块对预测结果做更加详细的分析。
print(classification_report(Y_test, y_predict))
```
